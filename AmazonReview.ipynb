```python
data = pd.read_csv('/content/drive/MyDrive/Dataset/AmazonReview.csv')
```

```python
# printing data
print("Data : ")
data.head()
```

```python
# Removing null values
data.dropna(axis=0, inplace=True)
```

```python
# Dividing target and feature column
x = data['Review']
y = data['Sentiment']
```

```python
x
```

```python
y
```

```python
# Sentence Tokenization
sent_tok = x.apply(lambda para: sent_tokenize(para, language="english"))
print("\nSentance Tokenization")
print(sent_tok)
```

```python
# Word Tokenization
word_tok = x.apply(lambda para: word_tokenize(para, language="english"))
print("\nWord Tokenization")
print(word_tok)
```

```python
# special symbols removal
x = x.apply(lambda elem: re.sub(r"[^a-zA-Z0-9]", " ", elem.lower()))
x
```

```python
# Stopwords List
stopwords_list = stopwords.words("english")


# print("\nStop-words : ")
print(stopwords_list)
print("Total Stop Words in corpus : ",len(stopwords_list))
```

```python
def stopWordsRemoval(sentance):
    new_x_sentance = [' '.join([word for word in sentance.split() if not word in stopwords_list])]
    return new_x_sentance


# Removal of Stopwords
x = x.apply(lambda para: stopWordsRemoval(para))
x
```

```python
# For Lemmatizing
word_lama = WordNetLemmatizer()

# Stemmed Data
def wordLematize(sentance):
    lematized_data = [' '.join([word_lama.lemmatize(word) for word in sentance[0].split()])]
    return lematized_data

# lametizing
x = x.apply(lambda para: wordLematize(para)[0])
x
```

```python
# Train-test split
xtrain, xtest, ytrain, ytest = train_test_split(x, y, test_size=0.20)
```

```python
# Applying Count Vectorize
# (Bag of Words)
dictonary_Words_bow = CountVectorizer()
bow_x = dictonary_Words_bow.fit_transform(xtrain).toarray()
bow_y = dictonary_Words_bow.transform(xtest).toarray()
```

```python
# Applying Tfidf Vectorizer (Tfidf Approach)
dictonary_Words_tfidf = TfidfVectorizer(stop_words="english", max_features=10000)
tfidf_x = dictonary_Words_tfidf.fit_transform(xtrain).toarray()
```

```python
# training the model
MB = MultinomialNB()
MB.fit(bow_x, ytrain)
```

```python
# predicting test values
print("The Predicted Values of Y : ")
ypred = MB.predict(bow_y)
print(ypred)
```

```python
# Accuracy Score
print("\nAccuracy Score", accuracy_score(ytest, ypred)*100)
```

```python
# Classification Report
print("Classification Report")
print(classification_report(y_true=ytest, y_pred=ypred))
```

```python
# training the model
rfc = RandomForestClassifier()
rfc.fit(bow_x, ytrain)
```

```python
# predicting test values
print("The Predicted Values of Y : ")
ypred = rfc.predict(bow_y)
print(ypred)
```

```python
# Accuracy Score
print("\nAccuracy Score", accuracy_score(ytest, ypred)*100)
```

```python
# Classification Report
print("Classification Report")
print(classification_report(y_true=ytest, y_pred=ypred))
```

```python
# training the model
dtc = DecisionTreeClassifier()
dtc.fit(bow_x, ytrain)
```
