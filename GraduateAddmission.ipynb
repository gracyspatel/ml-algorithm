```python
#  data splitting
features = data.iloc[:,:-1]
target = data.iloc[:,-1]
```

```python
# train test split
x_train,x_test,y_train,y_test = train_test_split(features,target,test_size=0.20)
```

```python
# Model training
linear_regression = LinearRegression()
linear_regression.fit(x_train,y_train)
```

```python
# predicting xtest
y_predicted = linear_regression.predict(x_test)
```

```python
# Performance
print("R2Score : ", r2_score(y_true=y_test,y_pred=y_predicted)*100)
print("Mean Squared Error",mean_squared_error(y_true=y_test,y_pred=y_predicted))
print("Root Mean Squared Error",math.sqrt(mean_squared_error(y_true=y_test,y_pred=y_predicted)))
```
